## Getting 5-days weather forecast

```python
# besh kunlik ob-havo malumotini olish uchun quyidagi kodni ishlatish mumkin:
url_5kun = "https://api.openweathermap.org/data/2.5/forecast"
response_5kun = requests.get(url_5kun, params=parametres)
data_5kun = response_5kun.json()
    
```

```python
# translating weather types to Uzbek
wtypes = {
    "Thunderstorm": "Bo'ron",
    "Drizzle": "Yengil yomg'ir",
    "Rain": "Yomg'ir",
    "Snow": "Qor",
    "Atmosphere": "Chaqmoq",
    "Clear": "Ochiq",
    "Clouds": "Bulutli"
}
```

```python
from datetime import datetime

def display_formatted_forecast(data_5kun, wtypes):
    print("\n📅 5-KUNLIK OB-HAVO MA'LUMOTI:\n")
    for i in range(7, 40, 8):
        dates = data_5kun['list'][i]
        kun = datetime.strptime(dates['dt_txt'][0:10], '%Y-%m-%d').strftime('%d-%B')
        shahar = data_5kun['city']['name']
        daraja = int(dates['main']['temp'])
        weather = dates['weather'][0]['main']
        havo = wtypes[weather]
        
        # Add emojis based on weather
        weather_emoji = {
            "Clear": "☀️",
            "Clouds": "☁️",
            "Rain": "🌧️",
            "Snow": "❄️",
            "Thunderstorm": "⛈️",
            "Drizzle": "🌦️"
        }.get(weather, "🌈")
        
        print(f"{kun}: {shahar}da {daraja}°C | {weather_emoji}{havo}")
```
